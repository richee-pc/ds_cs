## 0. 준비 &mdash; 반드시 먼저 실행하세요


```python
# 정렬 학습지에 나온 배열입니다. 원하는 것으로 바꿔서 실행해 보세요.
data1 = [7, 4, 9, 6, 3, 8, 7, 5]          # 학습지 1번
data2 = [71, 49, 92, 55, 38, 82, 72, 53]  # 학습지 2번
data3 = [4, 5, 9, 3, 6, 8, 7, 1]          # 학습지 3번 (피벗 비교용)
radix1 = [8, 2, 7, 3, 5]                  # 슬라이드 기수정렬 예제 (한 자리)
radix2 = [28, 93, 39, 81, 62, 72, 38, 26] # 슬라이드 기수정렬 예제 (두 자리)

# 비교 횟수와 이동 횟수를 세기 위한 상자
count = {"cmp": 0, "move": 0}

def reset():
    count["cmp"] = 0
    count["move"] = 0

def report(name, before, after):
    print(f"[{name}]")
    print(f"  처음 : {before}")
    print(f"  결과 : {after}")
    print(f"  비교 {count['cmp']}번,  이동(교환) {count['move']}번")

print("준비 완료! 아래 칸들을 순서대로 실행해 보세요.")
```

## 1. 선택 정렬

제일 작은 값을 골라 앞자리와 바꾸는 것을 반복합니다.
**교환 횟수가 가장 적은** 정렬입니다. 몇 번인지 세어 보세요.


```python
# ── 선택 정렬 (SELECTION SORT) ─────────────────────────────
# 수업 슬라이드 C 코드를 그대로 파이썬으로 옮긴 것입니다.
#
#   for(i=0; i<n-1; i++) {
#       least = i;
#       for(j=i+1; j<n; j++)          // 최솟값 탐색
#           if(list[j]<list[least]) least = j;
#       SWAP(list[i], list[least], temp);
#   }

def selection_sort(lst, show=True):
    a = lst[:]
    n = len(a)
    for i in range(n - 1):
        least = i
        for j in range(i + 1, n):     # 최솟값 탐색
            count["cmp"] += 1
            if a[j] < a[least]:
                least = j
        if least != i:
            a[i], a[least] = a[least], a[i]   # SWAP
            count["move"] += 1
        if show:
            print(f"  {i}번 자리 확정 → {a}")
    return a

reset()
before = data1[:]
after = selection_sort(before)
report("선택 정렬", data1, after)
```

## 2. 삽입 정렬

카드를 손에 쥐고 알맞은 자리에 끼워 넣듯 정렬합니다.
`data3` 처럼 **앞부분이 이미 정렬된 자료**를 넣으면 이동 횟수가 확 줄어듭니다. 바꿔서 해 보세요.


```python
# ── 삽입 정렬 (INSERTION SORT) ─────────────────────────────
#   for(i=1; i<n; i++){
#       key = list[i];
#       for(j=i-1; j>=0 && list[j]>key; j--)
#           list[j+1] = list[j];
#       list[j+1] = key;
#   }

def insertion_sort(lst, show=True):
    a = lst[:]
    for i in range(1, len(a)):
        key = a[i]
        j = i - 1
        while j >= 0 and a[j] > key:
            count["cmp"] += 1
            a[j + 1] = a[j]           # 한 칸 뒤로 밀기
            count["move"] += 1
            j -= 1
        if j >= 0:
            count["cmp"] += 1
        a[j + 1] = key
        if show:
            print(f"  {key} 를 {j+1}번 자리에 삽입 → {a}")
    return a

reset()
after = insertion_sort(data1)
report("삽입 정렬", data1, after)
```

## 3. 버블 정렬

옆 사람과만 비교해서 큰 값을 뒤로 보냅니다.
선택 정렬과 **비교 횟수는 같은데 이동 횟수가 훨씬 많은** 것을 확인하세요.


```python
# ── 버블 정렬 (BUBBLE SORT) ────────────────────────────────
#   for(i=n-1; i>0; i--){
#       for(j=0; j<i; j++)
#           if(list[j]>list[j+1])
#               SWAP(list[j], list[j+1], temp);
#   }

def bubble_sort(lst, show=True):
    a = lst[:]
    n = len(a)
    for i in range(n - 1, 0, -1):
        for j in range(i):
            count["cmp"] += 1
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]   # SWAP
                count["move"] += 1
        if show:
            print(f"  {i}번 자리에 가장 큰 값 확정 → {a}")
    return a

reset()
after = bubble_sort(data1)
report("버블 정렬", data1, after)
```

## 4. 셸 정렬

간격(gap)을 두고 부분 리스트를 삽입 정렬합니다. gap이 절반씩 줄어듭니다.
삽입 정렬과 이동 횟수를 비교해 보세요.


```python
# ── 셸 정렬 (SHELL SORT) ───────────────────────────────────
#   inc_insertion_sort(list, first, last, gap)  : gap칸씩 건너뛰는 삽입 정렬
#   for( gap=n/2; gap>0; gap=gap/2 ) { if((gap%2)==0) gap++; ... }

def inc_insertion_sort(a, first, last, gap):
    for i in range(first + gap, last + 1, gap):
        key = a[i]
        j = i - gap
        while j >= first and key < a[j]:
            count["cmp"] += 1
            a[j + gap] = a[j]
            count["move"] += 1
            j -= gap
        if j >= first:
            count["cmp"] += 1
        a[j + gap] = key

def shell_sort(lst, show=True):
    a = lst[:]
    n = len(a)
    gap = n // 2
    while gap > 0:
        if gap % 2 == 0:
            gap += 1
        for i in range(gap):          # 부분 리스트의 개수는 gap
            inc_insertion_sort(a, i, n - 1, gap)
        if show:
            print(f"  gap={gap} 마친 뒤 → {a}")
        gap = gap // 2
    return a

reset()
after = shell_sort(data1)
report("셸 정렬", data1, after)
```

## 5. 퀵 정렬

피벗을 기준으로 좌우로 갈라 세우고, 양쪽을 다시 퀵 정렬합니다.

> **참고** &mdash; 수업 슬라이드의 `partition` 은 `} while(low<high);` 까지만 실려 있습니다.
> 코드 안에 `← 슬라이드에서 잘린 부분` 이라고 표시한 두 줄이 그 뒤에 이어지는 마무리입니다.


```python
# ── 퀵 정렬 (QUICK SORT) ───────────────────────────────────
# 수업 슬라이드의 partition 은 } while(low<high); 까지만 실려 있습니다.
# 아래 표시한 두 줄이 그 뒤에 이어지는 마무리입니다.

def partition(a, left, right, show=True):
    pivot = a[left]
    low = left
    high = right + 1
    while True:
        while True:                        # do low++ while(list[low]<pivot)
            low += 1
            if low > right:
                break
            count["cmp"] += 1
            if a[low] >= pivot:
                break
        while True:                        # do high-- while(list[high]>pivot)
            high -= 1
            if high < left:
                break
            count["cmp"] += 1
            if a[high] <= pivot:
                break
        if low < high:
            a[low], a[high] = a[high], a[low]
            count["move"] += 1
        else:
            break

    a[left], a[high] = a[high], a[left]     # ← 슬라이드에서 잘린 부분 ①
    count["move"] += 1
    if show:
        print(f"  피벗 {pivot} 를 {high}번 자리에 확정 → {a}")
    return high                             # ← 슬라이드에서 잘린 부분 ②

def quick_sort(a, left, right, show=True):
    if left < right:
        q = partition(a, left, right, show)
        quick_sort(a, left, q - 1, show)
        quick_sort(a, q + 1, right, show)
    return a

reset()
work = data1[:]
after = quick_sort(work, 0, len(work) - 1)
report("퀵 정렬", data1, after)
```

## 6. 학습지 3번 &mdash; 피벗 선택 방법 비교

같은 배열이라도 **피벗을 어떻게 고르느냐**에 따라 비교 횟수가 달라집니다.
특히 **이미 정렬된 배열**에 맨 왼쪽을 피벗으로 쓰면 어떻게 되는지 꼭 보세요.


```python
# ── 학습지 3번 : 피벗을 다르게 골라 보기 ───────────────────
# 같은 배열이라도 피벗을 어떻게 고르느냐에 따라 성능이 달라집니다.

def quick_with_pivot(lst, how):
    """how : 'left'  = 맨 왼쪽을 피벗으로
             'mid3'  = 왼쪽.중간.오른쪽 세 값의 중간값을 피벗으로"""
    a = lst[:]

    def qs(left, right):
        if left >= right:
            return
        if how == "mid3":
            mid = (left + right) // 2
            trio = sorted([(a[left], left), (a[mid], mid), (a[right], right)])
            m = trio[1][1]                       # 세 값 중 가운데 값의 위치
            a[left], a[m] = a[m], a[left]        # 그 값을 맨 왼쪽으로
        q = partition(a, left, right, show=False)
        qs(left, q - 1)
        qs(q + 1, right)

    qs(0, len(a) - 1)
    return a

for name, how, src in [("맨 왼쪽을 피벗", "left", data3),
                       ("세 값의 중간을 피벗", "mid3", data3),
                       ("맨 왼쪽 · 이미 정렬된 배열", "left", sorted(data3)),
                       ("세 값의 중간 · 이미 정렬된 배열", "mid3", sorted(data3))]:
    reset()
    out = quick_with_pivot(src, how)
    print(f"{name:28s} 입력 {src}")
    print(f"{'':28s} 비교 {count['cmp']:3d}번, 이동 {count['move']:2d}번  →  {out}")
    print()
```

## 7. 기수 정렬 &mdash; 줄기와 잎 그래프

값을 **한 번도 비교하지 않고** 정렬합니다.
자릿수만 보고 0~9번 줄기에 매달았다가 위에서부터 훑어 내립니다.
출력되는 줄기와 잎 그림이 수업 슬라이드의 그림과 같은지 확인하세요.


```python
# ── 기수 정렬 (RADIX SORT) — 줄기와 잎 그래프 방식 ─────────
# 값끼리 비교하지 않습니다. 자릿수만 보고 0~9번 줄기에 매달았다가
# 0번 줄기부터 차례로 훑어 내립니다.

def radix_sort(lst, show=True):
    a = lst[:]
    digits = len(str(max(a)))
    factor = 1
    for d in range(digits):
        name = "1의 자리" if d == 0 else f"{factor}의 자리"
        bucket = [[] for _ in range(10)]
        for v in a:
            bucket[(v // factor) % 10].append(v)
            count["move"] += 1

        if show:
            print(f"\n[{name}로 분류]  ← 줄기와 잎 그래프")
            for i in range(10):
                leaf = "  ".join(str(x) for x in bucket[i])
                print(f"   {i} │ {leaf}")

        a = [v for b in bucket for v in b]
        if show:
            print(f"   → 0번 줄기부터 훑어 내리면 : {a}")
        factor *= 10
    return a

reset()
print("=== 한 자리 예제 (슬라이드 p28) ===")
after1 = radix_sort(radix1)
report("기수 정렬 ①", radix1, after1)

print("\n=== 두 자리 예제 (슬라이드 p29) ===")
reset()
after2 = radix_sort(radix2)
report("기수 정렬 ②", radix2, after2)
```

## 8. 한눈에 비교하기

같은 자료를 여섯 가지 정렬로 각각 처리하고 비교/이동 횟수를 견주어 봅니다.


```python
# ── 다섯 가지 정렬을 한 번에 비교하기 ──────────────────────
# 같은 배열을 각각 정렬해서 비교 횟수와 이동 횟수를 견주어 봅니다.

def run(name, fn, src):
    reset()
    out = fn(src, show=False)
    ok = "O" if out == sorted(src) else "X"
    print(f"{name:10s} | 비교 {count['cmp']:4d} | 이동 {count['move']:4d} | 정렬성공 {ok}")

target = data1        # data2, data3, radix2 로 바꿔서도 해 보세요
print(f"자료 : {target}\n")
print(f"{'정렬':10s} | {'비교':>6s} | {'이동':>6s} |")
print("-" * 42)

run("선택 정렬", selection_sort, target)
run("삽입 정렬", insertion_sort, target)
run("버블 정렬", bubble_sort, target)
run("셸 정렬", shell_sort, target)

reset()
w = target[:]
quick_sort(w, 0, len(w) - 1, show=False)
print(f"{'퀵 정렬':10s} | 비교 {count['cmp']:4d} | 이동 {count['move']:4d} | 정렬성공 "
      f"{'O' if w == sorted(target) else 'X'}")

run("기수 정렬", radix_sort, target)

print("\n생각해 보기")
print(" - 비교 횟수가 가장 적은 정렬은? 이동 횟수가 가장 적은 정렬은?")
print(" - 선택 정렬과 버블 정렬의 비교 횟수는 왜 같을까?")
print(" - 기수 정렬의 비교 횟수가 0인 이유는?")
```

## 정리

| 정렬 | 평균 | 특징 |
|---|---|---|
| 선택 | O(n²) | 교환이 가장 적다 |
| 삽입 | O(n²) | 거의 정렬된 자료에 가장 유리 |
| 버블 | O(n²) | 이동이 가장 많다 |
| 셸 | O(n^1.5) | 간격을 두어 미리 대충 정리 |
| 퀵 | O(n log n) | 평균 최고, 최악은 O(n²) |
| 합병 | O(n log n) | 최악에도 보장, 메모리 추가 |
| 기수 | O(d·n) | 비교를 하지 않는다 |

> **모든 경우에 최적인 정렬 알고리즘은 없습니다.**
> 고르는 기준은 **비교 횟수**와 **이동 횟수** 두 가지입니다.
