## 코드 A · 순차 탐색

앞에서부터 하나씩 확인하는 방법입니다. 몇 번 만에 찾는지 세어 보세요.

**할 일** — `data` 에 내 주제의 데이터를 10개 이상 넣고, `target` 에 찾을 값을 넣은 뒤 실행하세요.


````python
# =========================================================
#  코드 A · 순차 탐색 - 앞에서부터 하나씩 확인하는 과정
# =========================================================

# [1] 내 주제의 데이터를 여기에 넣으세요 (10개 이상)
data = ["강민준", "김서연", "박도윤", "이하은", "정시우",
        "최지아", "한예준", "황유나", "오시윤", "윤채원"]

# [2] 찾을 값을 정하세요
target = "정시우"

# ---------- 아래는 고치지 않아도 됩니다 ----------
def sequential_search(data, target):
    print(f"[순차 탐색] '{target}' 찾기 - 전체 {len(data)}개\n")
    count = 0
    for i, item in enumerate(data):
        count += 1
        left = len(data) - i - 1
        print(f"  {count}단계 | 비교: {item} | 남은 범위: {left}개", end="")
        if item == target:
            print("  -> 찾음!")
            print(f"\n결과: 위치 {i}번, 비교 횟수 {count}번")
            return count
        print("  -> 다름")
    print(f"\n결과: 자료에 없음, 비교 횟수 {count}번")
    return count

sequential_search(data, target)
````

## 코드 B · 이진 탐색

정렬된 자료에서 **탐색 범위를 절반씩 줄이는** 방법입니다.
코드 A와 같은 데이터를 넣고 실행해서, 비교 횟수가 몇 번으로 줄어드는지 확인하세요.

> 이진 탐색은 **정렬이 반드시 필요**합니다. 코드 안에서 자동으로 정렬한 뒤 탐색합니다.


````python
# =========================================================
#  코드 B · 이진 탐색 - 탐색 범위를 절반씩 줄이는 과정
# =========================================================

# [1] 내 주제의 데이터를 여기에 넣으세요 (코드 A와 같은 데이터)
data = ["강민준", "김서연", "박도윤", "이하은", "정시우",
        "최지아", "한예준", "황유나", "오시윤", "윤채원"]

# [2] 찾을 값을 정하세요
target = "정시우"

# ---------- 아래는 고치지 않아도 됩니다 ----------
data = sorted(data)          # 이진 탐색은 정렬이 반드시 필요합니다
print("정렬한 자료:")
for i, v in enumerate(data):
    print(f"  [{i}] {v}")
print()

def binary_search(data, target):
    low, high, count = 0, len(data) - 1, 0
    print(f"[이진 탐색] '{target}' 찾기 - 전체 {len(data)}개\n")
    while low <= high:
        mid = (low + high) // 2
        count += 1
        print(f"  {count}단계 | 범위 [{low}~{high}] ({high-low+1}개)"
              f" | 가운데 [{mid}] '{data[mid]}' 와 비교", end="")
        if data[mid] == target:
            print("  -> 찾음!")
            print(f"\n결과: 위치 {mid}번, 비교 횟수 {count}번")
            return count
        elif data[mid] < target:
            print("  -> 목표가 더 큼, 오른쪽 절반만 남김")
            low = mid + 1
        else:
            print("  -> 목표가 더 작음, 왼쪽 절반만 남김")
            high = mid - 1
    print(f"\n결과: 자료에 없음, 비교 횟수 {count}번")
    return count

binary_search(data, target)
````

## 코드 C · 자료가 커지면 얼마나 차이 날까

자료 크기를 10부터 100만까지 늘려 가며 **평균 비교 횟수를 실제로 측정**합니다. 몇 초 걸릴 수 있습니다.

**볼 것** — 맨 오른쪽 `log2(n)` 값과 이진 탐색의 비교 횟수를 견주어 보세요. 거의 같을 것입니다.


````python
# =========================================================
#  코드 C · 자료 크기별 비교 횟수 실측 실험
#  (마지막 100만 건은 몇 초 걸립니다. 기다려 주세요)
# =========================================================
import random, math

def seq_count(data, target):
    for i, v in enumerate(data):
        if v == target:
            return i + 1
    return len(data)

def bin_count(data, target):
    low, high, c = 0, len(data) - 1, 0
    while low <= high:
        mid = (low + high) // 2
        c += 1
        if data[mid] == target:
            return c
        elif data[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return c

TRIALS = 20      # 무작위로 20번 찾아본 평균을 냅니다
print(f"{'자료 크기':>12} | {'순차 탐색':>10} | {'이진 탐색':>10} | {'몇 배':>8} | {'log2(n)':>8}")
print("-" * 62)

for n in [10, 100, 1000, 10000, 100000, 1000000]:
    data = list(range(n))
    s = b = 0
    for _ in range(TRIALS):
        t = random.randrange(n)
        s += seq_count(data, t)
        b += bin_count(data, t)
    s //= TRIALS
    b //= TRIALS
    ratio = s / b if b else 0
    print(f"{n:>12,} | {s:>10,} | {b:>10,} | {ratio:>7.1f}배 | {math.log2(n):>8.1f}")

print("\n이진 탐색의 비교 횟수와 log2(n) 값을 비교해 보세요.")
````
